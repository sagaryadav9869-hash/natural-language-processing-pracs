```python
# Import required libraries
import nltk
import requests
from bs4 import BeautifulSoup

# Download necessary NLTK data
nltk.download('punkt')

# -------------------------------
# USER INPUT
# -------------------------------
text_input = input("Enter English or Hindi text: ")

# Sentence Tokenization
sentences = nltk.sent_tokenize(text_input)

# Word Tokenization
words = nltk.word_tokenize(text_input)

print("\nSentence Tokens:")
print(sentences)

print("\nWord Tokens:")
print(words)

# -------------------------------
# SCRIPT VALIDATION
# -------------------------------

print("\nScript Validation:")

for word in words:
    if any('\u0900' <= ch <= '\u097F' for ch in word):
        print(word, "-> Hindi Script")
    elif word.isalpha():
        print(word, "-> English Script")

# -------------------------------
# TEXT FILE INPUT
# -------------------------------

file = open("sample.txt", "r", encoding="utf-8")
file_text = file.read()
file.close()

print("\nText from File:")
print(file_text)

print("\nFile Word Tokens:")
print(nltk.word_tokenize(file_text))

# -------------------------------
# WEB SCRAPING
# -------------------------------

url = input("\nEnter URL for scraping: ")
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

web_text = soup.get_text()

print("\nScraped Text Tokens:")
print(nltk.word_tokenize(web_text[:500]))
```
