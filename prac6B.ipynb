```python
# Import required libraries
import nltk
from nltk.corpus import wordnet

# Download WordNet
nltk.download('wordnet')
nltk.download('omw-1.4')

# -------------------------
# USER INPUT
# -------------------------
word = input("Enter a word: ")

# -------------------------
# SYNSETS
# -------------------------
synsets = wordnet.synsets(word)

print("\nSynsets:")
for syn in synsets:
    print(syn.name(), "->", syn.definition())

# -------------------------
# SYNONYMS & ANTONYMS
# -------------------------
synonyms = []
antonyms = []

for syn in synsets:
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())
        if lemma.antonyms():
            antonyms.append(lemma.antonyms()[0].name())

print("\nSynonyms:")
print(set(synonyms))

print("\nAntonyms:")
print(set(antonyms))

# -------------------------
# HYPERNYMS
# -------------------------
print("\nHypernyms:")
for syn in synsets:
    for hyper in syn.hypernyms():
        print(hyper.name())

# -------------------------
# HYPONYMS
# -------------------------
print("\nHyponyms:")
for syn in synsets:
    for hypo in syn.hyponyms():
        print(hypo.name())
```
