```python
# Import required libraries
import nltk
from nltk.tokenize import word_tokenize
from nltk.util import ngrams

# Download required resource
nltk.download('punkt_tab')

# --------------------------
# USER INPUT
# --------------------------
text = input("Enter text: ")

# --------------------------
# TOKENIZATION
# --------------------------
tokens = word_tokenize(text)

print("\nTokens:")
print(tokens)

# --------------------------
# UNIGRAM
# --------------------------
print("\nUnigrams:")
for unigram in ngrams(tokens, 1):
    print(unigram)

# --------------------------
# BIGRAM
# --------------------------
print("\nBigrams:")
for bigram in ngrams(tokens, 2):
    print(bigram)

# --------------------------
# TRIGRAM
# --------------------------
print("\nTrigrams:")
for trigram in ngrams(tokens, 3):
    print(trigram)
```
