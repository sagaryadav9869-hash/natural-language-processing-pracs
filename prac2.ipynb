```python
# Import required libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required resources
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# -------------------------
# USER INPUT
# -------------------------
text = input("Enter English or Hindi text: ")

# -------------------------
# TOKENIZATION
# -------------------------
tokens = word_tokenize(text)

# -------------------------
# STOPWORDS
# -------------------------
english_stopwords = set(stopwords.words('english'))

# Basic Hindi stopwords (manual)
hindi_stopwords = {'और', 'है', 'का', 'के', 'की', 'में', 'यह', 'से'}

print("\nStopwords Found:")

for word in tokens:
    if word.lower() in english_stopwords or word in hindi_stopwords:
        print(word)

# -------------------------
# STEMMING
# -------------------------
stemmer = PorterStemmer()

print("\nStemming Output:")

for word in tokens:
    if word.isalpha():
        print(word, "->", stemmer.stem(word))

# -------------------------
# LEMMATIZATION
# -------------------------
lemmatizer = WordNetLemmatizer()

print("\nLemmatization Output:")

for word in tokens:
    if word.isalpha():
        print(word, "->", lemmatizer.lemmatize(word))
```
