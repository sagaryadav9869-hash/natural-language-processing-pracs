```python
# Import required libraries
import nltk
from nltk.corpus import brown
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.util import ngrams
import random

# Download required resources
nltk.download('brown')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')

# ---------------------------------------
# LOAD TEXT FROM BROWN CORPUS
# ---------------------------------------
categories = brown.categories()

print("Available Brown Corpus Categories:")
print(categories)

category_input = input("\nEnter a category from above list: ")

words = brown.words(categories=category_input)

# Take first 100 words for analysis
sample_words = words[:100]

print("\nSample Words from Brown Corpus:")
print(sample_words)

# ---------------------------------------
# MORPHOLOGICAL ANALYSIS
# ---------------------------------------

# POS Tagging
pos_tags = pos_tag(sample_words)

print("\nPOS Tagging:")
for word, tag in pos_tags[:20]:
    print(word, "->", tag)

# Lemmatization
lemmatizer = WordNetLemmatizer()

print("\nLemmatization:")
for word in sample_words[:20]:
    if word.isalpha():
        print(word, "->", lemmatizer.lemmatize(word))

# Prefix and Suffix Identification
print("\nPrefix and Suffix:")
for word in sample_words[:20]:
    if len(word) > 4 and word.isalpha():
        print(word, "Prefix:", word[:2], "Suffix:", word[-2:])

# ---------------------------------------
# TEXT GENERATION USING BIGRAM
# ---------------------------------------

print("\nText Generation (Bigram Based):")

bigrams = list(ngrams(sample_words, 2))

random_bigram = random.choice(bigrams)

print("Generated Text:")
print(random_bigram[0], random_bigram[1])

# ---------------------------------------
# WORD GENERATION
# ---------------------------------------

root_word = input("\nEnter a root word for word generation: ")

print("Generated Forms:")
print(root_word)
print(root_word + "s")
print(root_word + "ed")
print(root_word + "ing")
```
