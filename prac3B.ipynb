```python
# Import required libraries
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download required resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# ------------------------------------
# USER INPUT
# ------------------------------------
text = input("Enter text: ")

# ------------------------------------
# TOKENIZATION
# ------------------------------------
tokens = word_tokenize(text)

# ------------------------------------
# STOPWORDS REMOVAL
# ------------------------------------
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in tokens:
    if word.lower() not in stop_words and word.isalpha():
        filtered_words.append(word)

print("\nFiltered Words:")
print(filtered_words)

# ------------------------------------
# MORPHOLOGICAL ANALYSIS
# ------------------------------------
lemmatizer = WordNetLemmatizer()

print("\nPrefix | Base Word | Suffix")

for word in filtered_words:
    lemma = lemmatizer.lemmatize(word)
    
    # Extract prefix (first 2 letters)
    prefix = word[:2] if len(word) > 2 else word
    
    # Extract suffix (last 2 letters)
    suffix = word[-2:] if len(word) > 2 else word
    
    print(word, "->", prefix, "|", lemma, "|", suffix)

# ------------------------------------
# WORD GENERATION
# ------------------------------------
root_word = input("\nEnter base word for generation: ")

print("\nGenerated Words:")
print(root_word)
print(root_word + "s")
print(root_word + "ed")
print(root_word + "ing")
print(root_word + "er")
```
